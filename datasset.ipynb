```python
x = my_data_raw.column("Daily_Social_Media_Hours")
y = my_data_raw.column("Mental_Health_Score")

plt.Figure(figsize=(8, 6))
plt.scatter(x, y, s=8, alpha=0.15)

m, b = np.polyfit(x, y, 1)
plt.plot(x, m*x + b, linewidth=2, color="red")

plt.xlabel("Daily Social Media Hours")
plt.ylabel("Mental Health Score")
plt.title("Social Media Usage vs Mental Health")

plt.show()
```
